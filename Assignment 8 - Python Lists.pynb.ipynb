```python
##Q1. Accessing List Elements
##Task: List ka first, middle aur last element nikalna hai, original list change kiye baghair.

numbers = [10, 20, 30, 40, 50, 60, 70]

first = numbers[0]
middle = numbers[len(numbers) // 2]
last = numbers[-1]

print(first)
print(middle)
print(last)
```

```python
## Q2. List Slicing Task: First five elements, last four elements aur reverse order mein har doosra element nikalna hai.
numbers = [5, 10, 15, 20, 25, 30, 35, 40, 45, 50]

first_five = numbers[:5]
last_four = numbers[-4:]
every_second_reverse = numbers[::-2]

print(first_five)
print(last_four)
print(every_second_reverse)
```

```python
matrix = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]

main_diagonal = [matrix[0][0], matrix[1][1], matrix[2][2]]
secondary_diagonal = [matrix[0][2], matrix[1][1], matrix[2][0]]
middle_column = [matrix[0][1], matrix[1][1], matrix[2][1]]

print(main_diagonal, sum(main_diagonal))
print(secondary_diagonal, sum(secondary_diagonal))
print(middle_column, sum(middle_column))
```
